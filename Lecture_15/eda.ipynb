````python
# Create figure and 2x2 axes array
fig, axes = plt.subplots(2, 2, figsize=(8, 8))

# Top-Left
axes[0, 0].plot([1, 2, 3, 4], [1, 4, 9, 16])
axes[0, 0].set_title("Line")

# Top-Right
axes[0, 1].scatter([1, 2, 3, 4], [4, 3, 2, 1])
axes[0, 1].set_title("Scatter")

# Bottom-Left
axes[1, 0].bar(["X", "Y", "Z"], [10, 15, 7])
axes[1, 0].set_title("Bar")

# Bottom-Right
axes[1, 1].hist(np.random.randn(100), bins=15, color="purple")
axes[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
````

````python

# Set aesthetic theme
sns.set_theme(style="whitegrid")

````

````python
df_weather.
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Scatter plot
sns.scatterplot(data=df_weather, x="Date", y="Temperature", ax=axes[0])
axes[0].set_title("Temp vs time")

# Plot 2: Box plot
sns.boxplot(data=df_weather, x="Date", y="Humidity", ax=axes[1])
axes[1].set_title("Humidity vs time")

plt.tight_layout()
plt.show()
````

````python
fig, axes = plt.subplots(2, 1, figsize=(8, 8))

# Plot 1: Histogram with KDE curve
sns.histplot(data=df_weather, x="Temperature", kde=True, ax=axes[0], color="teal")
axes[0].set_title("Temp Histogram")

# Plot 2: Bar plot
sns.barplot(data=df_weather, x="Date", y="Humidity", estimator=sum, ax=axes[1], palette="Blues")
axes[1].set_title("Humid bar")

plt.tight_layout()
plt.show()
````
